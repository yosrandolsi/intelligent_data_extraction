```python
import re
from PIL import Image
import pytesseract

# Fonction pour extraire le texte d'une image à l'aide de pytesseract
def extract_text_from_image(image_path):
    image = Image.open(image_path)
    text = pytesseract.image_to_string(image)
    return text

# Fonction pour classifier le document en fonction de son contenu
def classify_document(text):
    # Expressions régulières pour identifier différents types de documents
    invoice_pattern = re.compile(r'invoice', re.IGNORECASE)
    cin_pattern = re.compile(r'CIN', re.IGNORECASE)
    receipt_pattern = re.compile(r'receipt', re.IGNORECASE)

    # Vérifier s'il s'agit d'une facture (invoice)
    if invoice_pattern.search(text):
        return "Invoice"
    # Vérifier s'il s'agit d'une carte d'identité nationale (CIN)
    elif cin_pattern.search(text):
        return "CIN"
    # Vérifier s'il s'agit d'un reçu (receipt)
    elif receipt_pattern.search(text):
        return "Receipt"
    else:
        return "Unknown"

# Chemin vers l'image
image_path = r"D:\Desktop\extraction2\invoice_extraction\docquery\Receipt-Templates-2.jpg"

# Extraire le texte de l'image
extracted_text = extract_text_from_image(image_path)

# Classifier le document en fonction du texte extrait
document_type = classify_document(extracted_text)

# Afficher le type de document
print("Document type:", document_type)
```

```python
from flask import Flask, request, jsonify
from flask_cors import CORS
from PIL import Image
import io
import base64
from transformers import pipeline
import fitz

app = Flask(__name__)
CORS(app)

# Chargement du modèle docQuery
nlp = pipeline(
    "document-question-answering",
    model="impira/layoutlm-document-qa",
)

@app.route('/extract-information', methods=['POST'])
def extract_information():
    if request.method == 'POST':
        # Vérifier si un fichier est téléchargé
        if 'document' not in request.files:
            return jsonify({'error': 'No document found in request'}), 400

        # Accéder au fichier téléchargé
        document_file = request.files['document']
        
        # Lire le contenu binaire du fichier PDF
        document_content = document_file.read()

        # Convertir le contenu binaire en un objet de flux en mémoire
        pdf_stream = io.BytesIO(document_content)

        # Définir les questions
        questions = {
            "invoice_number": "What is the invoice number?",
            "invoice_date": "What is the invoice date?",
            "total_amount": "What is the total amount?",
            "subtotal": "What is the subtotal?",
            "due_date": "What is the due date?",
            "vendor_name": "What is the vendor name?",
            # Ajoutez d'autres questions selon les informations que vous voulez extraire
        }

        result = {}
        # Poser les questions au modèle pour extraire les informations
        pdf_document = fitz.open(stream=pdf_stream)
        for page_number in range(len(pdf_document)):
            page = pdf_document.load_page(page_number)
            img_bytes = page.get_pixmap().tobytes()

            # Convertir l'image en chaîne base64
            buffered = io.BytesIO(img_bytes)
            img_pil = Image.open(buffered)
            buffered = io.BytesIO()
            img_pil.save(buffered, format="JPEG")
            img_base64 = base64.b64encode(buffered.getvalue()).decode('utf-8')

            # Poser les questions pour chaque page du PDF
            page_result = {}
            for key, question in questions.items():
                output = nlp(img_base64, question)
                if len(output) > 0:
                    page_result[key] = {'answer': output[0]['answer'], 'score': output[0]['score']}
                else:
                    page_result[key] = {'answer': "No answer found", 'score': None}
            result[f'page_{page_number+1}'] = page_result

        pdf_document.close()

        # Imprimer les informations d'extraction dans la console du serveur Flask
        print('Extraction result:', result)

        # Retourner les résultats
        return jsonify(result)

    else:
        return jsonify({'error': 'Invalid request method'}), 405

if __name__ == '__main__':
    app.run(debug=True)
```

```python
from transformers import pipeline

nlp = pipeline(
    "document-question-answering",
    model="impira/layoutlm-document-qa",
)

image_url = "D:\Desktop\DocQuery2\docquery\invoice.jpg"  # Replace with your image path

result = {}

# Champs d'information à extraire
champs = ["invoice_number", "invoice_date", "total_amount"]

# Fonctions d'extraction d'informations
def extract_with_question(champ):
  """Extrait l'information en posant une question spécifique."""
  question = f"{champ} :"
  output = nlp(image_url, question)
  if len(output) > 0:
    return output[0]['answer']
  else:
    return "Aucune réponse trouvée"

def extract_with_keywords(champ, keywords):
  """Extrait l'information en cherchant des mots-clés dans la réponse."""
  for mot in keywords:
    if mot in output[0]['answer']:
      return output[0]['answer'].split(mot)[1].strip()
  return "Aucune réponse trouvée"

# Boucle d'extraction et gestion d'erreurs
for champ in champs:
  try:
    # Essayez d'abord l'extraction basée sur la question
    result[champ] = extract_with_question(champ)
  except Exception as e:
    print(f"Erreur pour la question '{champ}' : {e}")
    result[champ] = "Erreur d'extraction"
    continue

  # Si la question échoue, essayez l'extraction basée sur des mots-clés
  if result[champ] == "Aucune réponse trouvée" or result[champ] == "Erreur d'extraction":
    mots_cles = {
        "invoice_number": ["numéro", "facture"],
        "invoice_date": ["date"],
        "total_amount": ["total", "montant"],
    }
    result[champ] = extract_with_keywords(champ, mots_cles[champ])

# Afficher le résultat sous forme JSON
print(result)
```

```python
from transformers import pipeline
import fitz
from PIL import Image

def convert_page_to_image(page):
  # Récupérer les dimensions de la page
  mat = fitz.Matrix(2, 2)  # Ajustez les dimensions si nécessaire
  pix = page.get_pixmap(matrix=mat)

  # Convertir l'image en objet PIL
  img_pil = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)

  return img_pil

def build_question(champ):
  return f"{champ} :"

def extract_answer(champ, output):
  if len(output) > 0:
    return output[0]['answer']
  else:
    return "Aucune réponse trouvée"

# Initialisation du pipeline de question-réponse sur images
nlp = pipeline(
    "document-question-answering",
    model="impira/layoutlm-document-qa",
)

# Chemin vers le fichier PDF
pdf_path = "D:\Desktop\DocQuery2\docquery\Invoice_PGITUNISIA_Num090_2024_1.pdf"

# Ouvrir le fichier PDF avec PyMuPDF
pdf_document = fitz.open(pdf_path)

# Définir les champs à extraire
champs = ["invoice_number", "invoice_date", "total_amount"]

# Dictionnaire pour stocker les résultats
resultats = {}

for page_number in range(len(pdf_document)):
  # Obtenir la page
  page = pdf_document.load_page(page_number)

  # Convertir la page en image
  img_pil = convert_page_to_image(page)

  # Extraire les informations pour chaque champ
  for champ in champs:
    question = build_question(champ)
    answer = extract_answer(champ, nlp(img_pil, question))
    resultats[f"{champ}_{page_number + 1}"] = answer

# Fermer le document PDF
pdf_document.close()

# Afficher les résultats
for key, value in resultats.items():
  print(f"{key}: {value}")
```

```python
import fitz
from PIL import Image
from transformers import pipeline

# Define the question-answering model
nlp = pipeline("document-question-answering", model="impira/layoutlm-document-qa")


def build_question(champ):
    """Construit une question simple à partir du nom du champ."""
    return f"{champ} :"


def extract_answer(champ, output):
    if len(output) > 0:
        return output[0]["answer"]
    else:
        return "Aucune réponse trouvée"


def extract_info_from_pdf_single_image(pdf_path, champs):
    """Extrait les informations des champs spécifiés d'un PDF en une seule image."""

    # Convertir le document PDF en une seule image
    img = fitz.open(pdf_path).get_pixmap(matrix=fitz.Matrix(2, 2))
    img_pil = Image.frombytes("RGB", [img.width, img.height], img.samples)

    # Extraire les informations
    results = {champ: extract_answer(champ, nlp(img_pil, build_question(champ)))
                for champ in champs}

    return results


# Champs à extraire
champs = ["invoice_number", "date_facture", "montant_total"]

# Exemple d'utilisation
pdf_path = "D:\Desktop\DocQuery2\docquery\Invoice_PGITUNISIA_Num090_2024_1.pdf"

# Extraire les informations
results = extract_info_from_pdf_single_image(pdf_path, champs)

# Afficher les résultats
for champ, valeur in results.items():
    print(f"{champ}: {valeur}")
```

```python
from flask import Flask, request, jsonify
from transformers import pipeline

app = Flask(__name__)

# Initialisation du modèle
nlp = pipeline(
    "document-question-answering",
    model="impira/layoutlm-document-qa",
)

# Route pour l'extraction des informations à partir de l'URL de l'image
@app.route('/extract-information', methods=['GET'])
def extract_information():
    try:
        # Récupérer l'URL de l'image depuis les paramètres de l'URL
        image_url = request.args.get('image_url')

        # Question et réponse pour chaque information à extraire
        questions = {
            "invoice_number": "What is the invoice number?",
            "invoice_date": "What is the invoice date?",
            "total_amount": "What is the total amount?",
            # Ajoutez d'autres questions selon les informations que vous voulez extraire
        }

        result = {}

        # Pour chaque question, obtenez la réponse en utilisant le pipeline et ajoutez-la au résultat
        for key, question in questions.items():
            output = nlp(image_url, question)
            # Vérifiez s'il y a une réponse et sélectionnez la première réponse
            if len(output) > 0:
                result[key] = output[0]['answer']
            else:
                result[key] = "No answer found"  # Gérer le cas où aucune réponse n'est trouvée

        # Retourner le résultat sous forme de réponse JSON
        return jsonify(result), 200
    except Exception as e:
        # En cas d'erreur, retourner un message d'erreur avec le code HTTP 500
        return jsonify({'error': 'Une erreur est survenue lors de l\'extraction des informations.'}), 500

if __name__ == '__main__':
    app.run(debug=True)
```
